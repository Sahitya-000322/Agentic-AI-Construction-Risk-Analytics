````python
import pandas as pd
import numpy as np

# Load Schedule Intelligence dataset
df = pd.read_csv(
    '/kaggle/input/datasets/sakshibhosale1904/schedule-intelligence-dataset/schdule_intelligence/delay_prediction.csv'
)

# Basic information
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nFirst 5 Rows:")
display(df.head())

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDelay Distribution:")
print(df['Delay'].value_counts())
````

````python
# ============================================
# SCHEDULE INTELLIGENCE — EDA
# ============================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 1. Dataset overview
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

# 2. Missing values
print("\nMissing Values:")
print(df.isnull().sum())

# 3. Duplicate rows
print("\nDuplicate Rows:", df.duplicated().sum())

# 4. Statistical summary
print("\nStatistical Summary:")
display(df.describe(include='all'))

# 5. Target distribution
print("\nDelay Distribution:")
print(df['Delay'].value_counts())

print("\nDelay Percentage:")
print(df['Delay'].value_counts(normalize=True) * 100)

# 6. Unique values for each column
print("\nUnique Values:")
for column in df.columns:
    print(f"{column}: {df[column].nunique()} unique values")

# 7. Numerical columns
numeric_columns = df.select_dtypes(include=np.number).columns

print("\nNumerical Columns:")
print(numeric_columns.tolist())

# 8. Correlation matrix
print("\nCorrelation Matrix:")
display(df[numeric_columns].corr())

# 9. Delay distribution plot
plt.figure(figsize=(6, 4))
df['Delay'].value_counts().plot(kind='bar')
plt.title('Schedule Delay Distribution')
plt.xlabel('Delay')
plt.ylabel('Number of Projects')
plt.xticks(rotation=0)
plt.show()
````

````python
# ============================================
# STEP 3 — FEATURE SELECTION & PREPROCESSING
# ============================================

# Select features
features = ['Weather', 'Workers', 'Budget', 'Progress']

X = df[features].copy()

# Convert target:
# No Delay = 0
# Delay = 1
y = df['Delay'].map({'No': 0, 'Yes': 1})

print("Selected Features:")
print(X.columns.tolist())

print("\nTarget Mapping:")
print("0 = No Delay")
print("1 = Delay")

print("\nFeature Data:")
display(X.head())

print("\nTarget Data:")
display(y.head())

# Check for missing values
print("\nMissing Values in Features:")
print(X.isnull().sum())

print("\nMissing Values in Target:")
print(y.isnull().sum())

# Check data types
print("\nFeature Data Types:")
print(X.dtypes)

# Final shape
print("\nFeatures Shape:", X.shape)
print("Target Shape:", y.shape)
````

````python
# ============================================
# STEP 4 — ENCODING & TRAIN/TEST SPLIT
# ============================================

from sklearn.model_selection import train_test_split

# Convert Weather into numerical values
X_encoded = pd.get_dummies(
    X,
    columns=['Weather'],
    dtype=int
)

print("Encoded Features:")
display(X_encoded.head())

print("\nEncoded Feature Names:")
print(X_encoded.columns.tolist())

# Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\n===================================")
print("TRAIN/TEST SPLIT")
print("===================================")

print("Training samples:", X_train.shape[0])
print("Testing samples :", X_test.shape[0])
print("Number of features:", X_train.shape[1])

print("\nTraining target distribution:")
print(y_train.value_counts())

print("\nTesting target distribution:")
print(y_test.value_counts())
````

````python
# ============================================
# STEP 5 — RANDOM FOREST
# ============================================

from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

# Create Random Forest model
rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

# Train
rf_model.fit(X_train, y_train)

# Predictions
rf_predictions = rf_model.predict(X_test)

# Evaluation
rf_accuracy = accuracy_score(y_test, rf_predictions)
rf_precision = precision_score(y_test, rf_predictions)
rf_recall = recall_score(y_test, rf_predictions)
rf_f1 = f1_score(y_test, rf_predictions)

print("===================================")
print("RANDOM FOREST RESULTS")
print("===================================")

print(f"Accuracy  : {rf_accuracy:.4f}")
print(f"Precision : {rf_precision:.4f}")
print(f"Recall    : {rf_recall:.4f}")
print(f"F1 Score  : {rf_f1:.4f}")

print("\nClassification Report:")
print(classification_report(
    y_test,
    rf_predictions,
    target_names=['No Delay', 'Delay']
))

print("Confusion Matrix:")
print(confusion_matrix(y_test, rf_predictions))
````

````python
# ============================================
# STEP 6 — GRADIENT BOOSTING
# ============================================

from sklearn.ensemble import GradientBoostingClassifier

# Create Gradient Boosting model
gb_model = GradientBoostingClassifier(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=3,
    random_state=42
)

# Train
gb_model.fit(X_train, y_train)

# Predictions
gb_predictions = gb_model.predict(X_test)

# Evaluation
gb_accuracy = accuracy_score(y_test, gb_predictions)
gb_precision = precision_score(y_test, gb_predictions)
gb_recall = recall_score(y_test, gb_predictions)
gb_f1 = f1_score(y_test, gb_predictions)

print("===================================")
print("GRADIENT BOOSTING RESULTS")
print("===================================")

print(f"Accuracy  : {gb_accuracy:.4f}")
print(f"Precision : {gb_precision:.4f}")
print(f"Recall    : {gb_recall:.4f}")
print(f"F1 Score  : {gb_f1:.4f}")

print("\nClassification Report:")
print(classification_report(
    y_test,
    gb_predictions,
    target_names=['No Delay', 'Delay']
))

print("Confusion Matrix:")
print(confusion_matrix(y_test, gb_predictions))
````

````python
# ============================================
# STEP 7 — XGBOOST
# ============================================

from xgboost import XGBClassifier

# Create XGBoost model
xgb_model = XGBClassifier(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=3,
    random_state=42,
    eval_metric='logloss'
)

# Train
xgb_model.fit(X_train, y_train)

# Predictions
xgb_predictions = xgb_model.predict(X_test)

# Evaluation
xgb_accuracy = accuracy_score(y_test, xgb_predictions)
xgb_precision = precision_score(y_test, xgb_predictions)
xgb_recall = recall_score(y_test, xgb_predictions)
xgb_f1 = f1_score(y_test, xgb_predictions)

print("===================================")
print("XGBOOST RESULTS")
print("===================================")

print(f"Accuracy  : {xgb_accuracy:.4f}")
print(f"Precision : {xgb_precision:.4f}")
print(f"Recall    : {xgb_recall:.4f}")
print(f"F1 Score  : {xgb_f1:.4f}")

print("\nClassification Report:")
print(classification_report(
    y_test,
    xgb_predictions,
    target_names=['No Delay', 'Delay']
))

print("Confusion Matrix:")
print(confusion_matrix(y_test, xgb_predictions))
````

````python
# ============================================
# STEP 8 — 5-FOLD CROSS-VALIDATION
# ============================================

from sklearn.model_selection import StratifiedKFold, cross_validate

# 5-Fold Stratified Cross-Validation
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

# Models
models = {
    "Random Forest": RandomForestClassifier(
        n_estimators=100,
        random_state=42
    ),

    "Gradient Boosting": GradientBoostingClassifier(
        n_estimators=100,
        learning_rate=0.1,
        max_depth=3,
        random_state=42
    ),

    "XGBoost": XGBClassifier(
        n_estimators=100,
        learning_rate=0.1,
        max_depth=3,
        random_state=42,
        eval_metric='logloss'
    )
}

# Scoring metrics
scoring = {
    'accuracy': 'accuracy',
    'precision': 'precision',
    'recall': 'recall',
    'f1': 'f1'
}

results = []

for name, model in models.items():

    scores = cross_validate(
        model,
        X_encoded,
        y,
        cv=cv,
        scoring=scoring
    )

    results.append({
        'Model': name,
        'CV Accuracy': scores['test_accuracy'].mean(),
        'CV Precision': scores['test_precision'].mean(),
        'CV Recall': scores['test_recall'].mean(),
        'CV F1': scores['test_f1'].mean()
    })

# Create comparison table
cv_results = pd.DataFrame(results)

# Round values
cv_results = cv_results.round(4)

print("==========================================")
print("5-FOLD CROSS-VALIDATION RESULTS")
print("==========================================")

display(cv_results)
````

````python
# ============================================
# STEP 9 — XGBOOST FEATURE IMPORTANCE
# ============================================

# Get feature importance
feature_importance = pd.DataFrame({
    'Feature': X_encoded.columns,
    'Importance': xgb_model.feature_importances_
})

# Sort from highest to lowest
feature_importance = feature_importance.sort_values(
    by='Importance',
    ascending=False
)

print("===================================")
print("XGBOOST FEATURE IMPORTANCE")
print("===================================")

display(feature_importance)

# Plot
plt.figure(figsize=(8, 5))

plt.bar(
    feature_importance['Feature'],
    feature_importance['Importance']
)

plt.title('XGBoost Feature Importance - Schedule Intelligence')
plt.xlabel('Features')
plt.ylabel('Importance')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
````

````python
# ============================================
# STEP 10 — FINAL MODEL TRAINING & SAVING
# ============================================

import joblib
from xgboost import XGBClassifier

# Create final XGBoost model
final_model = XGBClassifier(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=3,
    random_state=42,
    eval_metric='logloss'
)

# Train final model on training data
final_model.fit(X_train, y_train)

# Save model
model_path = '/kaggle/working/schedule_delay_prediction_model.pkl'

joblib.dump(final_model, model_path)

print("===================================")
print("FINAL MODEL SAVED SUCCESSFULLY!")
print("===================================")
print("Model:", model_path)
print("Model Type: XGBoost Classifier")
print("Purpose: Schedule Delay Prediction")
print("Features:", X_encoded.columns.tolist())
print("Target: Delay")
print("0 = No Delay")
print("1 = Delay")
````

````python
# ============================================
# STEP 3 — FEATURE SELECTION & PREPROCESSING
# ============================================

# Select input features
features_schedule = [
    'Planned_Duration',
    'Completed_Work',
    'Pending_Work'
]

X_schedule = df_schedule[features_schedule].copy()

# Target
y_schedule = df_schedule['Delay_Days'].copy()

print("Selected Features:")
print(features_schedule)

print("\nTarget:")
print("Delay_Days")

print("\nFeature Data:")
display(X_schedule.head())

print("\nTarget Data:")
display(y_schedule.head())

# Check missing values
print("\nMissing Values in Features:")
print(X_schedule.isnull().sum())

print("\nMissing Values in Target:")
print(y_schedule.isnull().sum())

# Check data types
print("\nFeature Data Types:")
print(X_schedule.dtypes)

print("\nTarget Data Type:")
print(y_schedule.dtype)

# Check target statistics
print("\nTarget Statistics:")
print(y_schedule.describe())

print("\nFeature Shape:", X_schedule.shape)
print("Target Shape:", y_schedule.shape)
````

````python
# ============================================
# STEP 4 — TRAIN/TEST SPLIT
# ============================================

from sklearn.model_selection import train_test_split

X_train_schedule, X_test_schedule, y_train_schedule, y_test_schedule = train_test_split(
    X_schedule,
    y_schedule,
    test_size=0.20,
    random_state=42
)

print("===================================")
print("TRAIN/TEST SPLIT")
print("===================================")

print("Training samples:", X_train_schedule.shape[0])
print("Testing samples :", X_test_schedule.shape[0])
print("Number of features:", X_train_schedule.shape[1])

print("\nFeature Names:")
print(X_train_schedule.columns.tolist())

print("\nTraining Target Shape:", y_train_schedule.shape)
print("Testing Target Shape :", y_test_schedule.shape)
````

````python
# ============================================
# STEP 5 — RANDOM FOREST REGRESSION
# ============================================

from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Create Random Forest Regressor
rf_schedule = RandomForestRegressor(
    n_estimators=200,
    random_state=42
)

# Train the model
rf_schedule.fit(X_train_schedule, y_train_schedule)

# Make predictions
y_pred_rf = rf_schedule.predict(X_test_schedule)

# Calculate evaluation metrics
mae_rf = mean_absolute_error(y_test_schedule, y_pred_rf)
rmse_rf = np.sqrt(mean_squared_error(y_test_schedule, y_pred_rf))
r2_rf = r2_score(y_test_schedule, y_pred_rf)

print("===================================")
print("RANDOM FOREST REGRESSION RESULTS")
print("===================================")

print(f"MAE  : {mae_rf:.4f}")
print(f"RMSE : {rmse_rf:.4f}")
print(f"R²   : {r2_rf:.4f}")

print("\nFirst 10 Actual vs Predicted Values:")
for actual, predicted in zip(y_test_schedule.iloc[:10], y_pred_rf[:10]):
    print(f"Actual: {actual:.2f}  |  Predicted: {predicted:.2f}")
````

````python
# ============================================
# STEP 6 — GRADIENT BOOSTING REGRESSION
# ============================================

from sklearn.ensemble import GradientBoostingRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Create Gradient Boosting Regressor
gb_schedule = GradientBoostingRegressor(
    n_estimators=200,
    learning_rate=0.05,
    max_depth=3,
    random_state=42
)

# Train the model
gb_schedule.fit(X_train_schedule, y_train_schedule)

# Make predictions
y_pred_gb = gb_schedule.predict(X_test_schedule)

# Evaluation metrics
mae_gb = mean_absolute_error(y_test_schedule, y_pred_gb)
rmse_gb = np.sqrt(mean_squared_error(y_test_schedule, y_pred_gb))
r2_gb = r2_score(y_test_schedule, y_pred_gb)

print("===================================")
print("GRADIENT BOOSTING REGRESSION RESULTS")
print("===================================")

print(f"MAE  : {mae_gb:.4f}")
print(f"RMSE : {rmse_gb:.4f}")
print(f"R²   : {r2_gb:.4f}")

print("\nFirst 10 Actual vs Predicted Values:")
for actual, predicted in zip(y_test_schedule.iloc[:10], y_pred_gb[:10]):
    print(f"Actual: {actual:.2f}  |  Predicted: {predicted:.2f}")
````

````python
# ============================================
# STEP 7 — XGBOOST REGRESSION
# ============================================

from xgboost import XGBRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Create XGBoost Regressor
xgb_schedule = XGBRegressor(
    n_estimators=200,
    learning_rate=0.05,
    max_depth=3,
    random_state=42,
    objective='reg:squarederror'
)

# Train the model
xgb_schedule.fit(X_train_schedule, y_train_schedule)

# Make predictions
y_pred_xgb = xgb_schedule.predict(X_test_schedule)

# Evaluation metrics
mae_xgb = mean_absolute_error(y_test_schedule, y_pred_xgb)
rmse_xgb = np.sqrt(mean_squared_error(y_test_schedule, y_pred_xgb))
r2_xgb = r2_score(y_test_schedule, y_pred_xgb)

print("===================================")
print("XGBOOST REGRESSION RESULTS")
print("===================================")

print(f"MAE  : {mae_xgb:.4f}")
print(f"RMSE : {rmse_xgb:.4f}")
print(f"R²   : {r2_xgb:.4f}")

print("\nFirst 10 Actual vs Predicted Values:")
for actual, predicted in zip(y_test_schedule.iloc[:10], y_pred_xgb[:10]):
    print(f"Actual: {actual:.2f}  |  Predicted: {predicted:.2f}")
````

````python
# ============================================
# STEP 8 — 5-FOLD CROSS-VALIDATION
# ============================================

from sklearn.model_selection import KFold, cross_validate
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from xgboost import XGBRegressor
import numpy as np

# 5-Fold Cross-Validation
kf = KFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

# Define models
models_schedule = {
    "Random Forest": RandomForestRegressor(
        n_estimators=200,
        random_state=42
    ),

    "Gradient Boosting": GradientBoostingRegressor(
        n_estimators=200,
        learning_rate=0.05,
        max_depth=3,
        random_state=42
    ),

    "XGBoost": XGBRegressor(
        n_estimators=200,
        learning_rate=0.05,
        max_depth=3,
        random_state=42,
        objective='reg:squarederror'
    )
}

print("==============================================")
print("SCHEDULE REGRESSION — 5-FOLD CROSS-VALIDATION")
print("==============================================")

for name, model in models_schedule.items():

    scores = cross_validate(
        model,
        X_schedule,
        y_schedule,
        cv=kf,
        scoring={
            'MAE': 'neg_mean_absolute_error',
            'RMSE': 'neg_root_mean_squared_error',
            'R2': 'r2'
        }
    )

    mae_scores = -scores['test_MAE']
    rmse_scores = -scores['test_RMSE']
    r2_scores = scores['test_R2']

    print(f"\n{name}")
    print("-" * 40)

    print(f"MAE  : {mae_scores.mean():.4f} ± {mae_scores.std():.4f}")
    print(f"RMSE : {rmse_scores.mean():.4f} ± {rmse_scores.std():.4f}")
    print(f"R²   : {r2_scores.mean():.4f} ± {r2_scores.std():.4f}")
````

````python
# ============================================
# STEP 1 — LOAD PROJECTS DATASET
# ============================================

import pandas as pd

projects_path = (
    "/kaggle/input/datasets/"
    "sakshibhosale1904/"
    "schedule-intelligence-dataset/"
    "schdule_intelligence/"
    "projects.csv"
)

df_projects = pd.read_csv(projects_path)

print("==============================================")
print("PROJECTS DATASET LOADED")
print("==============================================")

print("Dataset Shape:", df_projects.shape)

print("\nColumn Names:")
print(df_projects.columns.tolist())

print("\nFirst 10 Rows:")
display(df_projects.head(10))

print("\nData Types:")
print(df_projects.dtypes)

print("\nMissing Values:")
print(df_projects.isnull().sum())

print("\nBasic Statistics:")
display(df_projects.describe(include="all"))
````

````python
# ============================================
# STEP 3 — FEATURE SELECTION & PREPROCESSING
# ============================================

# Features
features_projects = [
    'Location',
    'Progress',
    'Budget',
    'Spent',
    'Workers'
]

X_projects = df_projects[features_projects].copy()

# Target
y_projects = df_projects['Delay'].map({
    'No': 0,
    'Yes': 1
})

print("==============================================")
print("PROJECT DATA — FEATURE SELECTION")
print("==============================================")

print("Selected Features:")
print(features_projects)

print("\nTarget: Delay")

print("\nFeature Data:")
display(X_projects.head())

print("\nTarget Distribution:")
print(y_projects.value_counts())
print("\nTarget Distribution (%):")
print((y_projects.value_counts(normalize=True) * 100).round(2))

print("\nMissing Values:")
print(X_projects.isnull().sum())

print("\nFeature Data Types:")
print(X_projects.dtypes)

print("\nFeature Shape:", X_projects.shape)
print("Target Shape:", y_projects.shape)
````

````python
# ============================================
# STEP 4 — ENCODE CATEGORICAL FEATURES
# ============================================

# One-hot encode Location
X_projects_encoded = pd.get_dummies(
    X_projects,
    columns=['Location'],
    dtype=int
)

print("==============================================")
print("PROJECT DATA — ENCODED FEATURES")
print("==============================================")

print("Original Features:")
print(features_projects)

print("\nEncoded Feature Names:")
print(X_projects_encoded.columns.tolist())

print("\nEncoded Data:")
display(X_projects_encoded.head())

print("\nFinal Feature Shape:", X_projects_encoded.shape)

print("\nData Types:")
print(X_projects_encoded.dtypes)
````

````python
# ============================================
# STEP 5 — TRAIN/TEST SPLIT
# ============================================

from sklearn.model_selection import train_test_split

X_train_projects, X_test_projects, y_train_projects, y_test_projects = train_test_split(
    X_projects_encoded,
    y_projects,
    test_size=0.20,
    random_state=42,
    stratify=y_projects
)

print("==============================================")
print("PROJECT DATA — TRAIN/TEST SPLIT")
print("==============================================")

print("Training samples:", X_train_projects.shape[0])
print("Testing samples :", X_test_projects.shape[0])
print("Number of features:", X_train_projects.shape[1])

print("\nTraining target distribution:")
print(y_train_projects.value_counts())

print("\nTesting target distribution:")
print(y_test_projects.value_counts())

print("\nTraining Feature Shape:", X_train_projects.shape)
print("Testing Feature Shape :", X_test_projects.shape)
````

````python
# ============================================
# STEP 6 — RANDOM FOREST CLASSIFICATION
# ============================================

from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

# Create Random Forest model
rf_projects = RandomForestClassifier(
    n_estimators=200,
    random_state=42
)

# Train
rf_projects.fit(
    X_train_projects,
    y_train_projects
)

# Predict
y_pred_rf_projects = rf_projects.predict(
    X_test_projects
)

# Evaluation
accuracy_rf_projects = accuracy_score(
    y_test_projects,
    y_pred_rf_projects
)

precision_rf_projects = precision_score(
    y_test_projects,
    y_pred_rf_projects
)

recall_rf_projects = recall_score(
    y_test_projects,
    y_pred_rf_projects
)

f1_rf_projects = f1_score(
    y_test_projects,
    y_pred_rf_projects
)

cm_rf_projects = confusion_matrix(
    y_test_projects,
    y_pred_rf_projects
)

print("==============================================")
print("RANDOM FOREST — PROJECT DELAY PREDICTION")
print("==============================================")

print(f"Accuracy  : {accuracy_rf_projects:.4f}")
print(f"Precision : {precision_rf_projects:.4f}")
print(f"Recall    : {recall_rf_projects:.4f}")
print(f"F1 Score  : {f1_rf_projects:.4f}")

print("\nConfusion Matrix:")
print(cm_rf_projects)
````

````python
# ============================================
# STEP 8 — XGBOOST CLASSIFICATION
# ============================================

from xgboost import XGBClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

# Create XGBoost model
xgb_projects = XGBClassifier(
    n_estimators=200,
    learning_rate=0.05,
    max_depth=3,
    random_state=42,
    eval_metric='logloss'
)

# Train
xgb_projects.fit(
    X_train_projects,
    y_train_projects
)

# Predict
y_pred_xgb_projects = xgb_projects.predict(
    X_test_projects
)

# Evaluation
accuracy_xgb_projects = accuracy_score(
    y_test_projects,
    y_pred_xgb_projects
)

precision_xgb_projects = precision_score(
    y_test_projects,
    y_pred_xgb_projects
)

recall_xgb_projects = recall_score(
    y_test_projects,
    y_pred_xgb_projects
)

f1_xgb_projects = f1_score(
    y_test_projects,
    y_pred_xgb_projects
)

cm_xgb_projects = confusion_matrix(
    y_test_projects,
    y_pred_xgb_projects
)

print("==============================================")
print("XGBOOST — PROJECT DELAY PREDICTION")
print("==============================================")

print(f"Accuracy  : {accuracy_xgb_projects:.4f}")
print(f"Precision : {precision_xgb_projects:.4f}")
print(f"Recall    : {recall_xgb_projects:.4f}")
print(f"F1 Score  : {f1_xgb_projects:.4f}")

print("\nConfusion Matrix:")
print(cm_xgb_projects)
````

````python
# ============================================
# STEP 10 — 5-FOLD CROSS-VALIDATION
# ============================================

from sklearn.model_selection import StratifiedKFold, cross_validate
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from xgboost import XGBClassifier
import numpy as np

# 5-Fold Stratified Cross-Validation
skf = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

# Define models
models_projects = {
    "Random Forest": RandomForestClassifier(
        n_estimators=200,
        random_state=42
    ),

    "Gradient Boosting": GradientBoostingClassifier(
        n_estimators=200,
        learning_rate=0.05,
        max_depth=3,
        random_state=42
    ),

    "XGBoost": XGBClassifier(
        n_estimators=200,
        learning_rate=0.05,
        max_depth=3,
        random_state=42,
        eval_metric='logloss'
    )
}

print("======================================================")
print("PROJECT DELAY — 5-FOLD CROSS-VALIDATION")
print("======================================================")

cv_results_projects = []

for name, model in models_projects.items():

    scores = cross_validate(
        model,
        X_projects_encoded,
        y_projects,
        cv=skf,
        scoring={
            'accuracy': 'accuracy',
            'precision': 'precision',
            'recall': 'recall',
            'f1': 'f1'
        }
    )

    accuracy_mean = scores['test_accuracy'].mean()
    accuracy_std = scores['test_accuracy'].std()

    precision_mean = scores['test_precision'].mean()
    precision_std = scores['test_precision'].std()

    recall_mean = scores['test_recall'].mean()
    recall_std = scores['test_recall'].std()

    f1_mean = scores['test_f1'].mean()
    f1_std = scores['test_f1'].std()

    cv_results_projects.append({
        "Model": name,
        "Accuracy": accuracy_mean,
        "Precision": precision_mean,
        "Recall": recall_mean,
        "F1 Score": f1_mean
    })

    print(f"\n{name}")
    print("-" * 45)
    print(f"Accuracy  : {accuracy_mean:.4f} ± {accuracy_std:.4f}")
    print(f"Precision : {precision_mean:.4f} ± {precision_std:.4f}")
    print(f"Recall    : {recall_mean:.4f} ± {recall_std:.4f}")
    print(f"F1 Score  : {f1_mean:.4f} ± {f1_std:.4f}")

# Create comparison table
cv_comparison_projects = pd.DataFrame(cv_results_projects)

print("\n======================================================")
print("CROSS-VALIDATION SUMMARY")
print("======================================================")

display(cv_comparison_projects.round(4))
````
